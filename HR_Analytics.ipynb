```python
# Data Analysis or Exploratory Data Analysis (EDA)
import pandas as pd
a = pd.read_csv('Employee Attrition.csv')

# Finding Total Number of Employees
b = a["EmployeeCount"].count()
print("*Total Number of employees are : ", b)

# Finding  Department Wise Number of Employees
a1=a.groupby("Department")["EmployeeCount"].count()
print("*Below are the Department Wise Number of Employees :\n",a1)

# Finding Which Department have maximum Number of Employees
a2=a.groupby("Department")["EmployeeCount"].count().idxmax()
a3=a.groupby("Department")["EmployeeCount"].count().max()
print("*Department which have maximum number of employees is :\n",a2,"\n*Number of employees of that Department are :\n",a3)

# Finding Which Department have minimum Number of Employees
a4=a.groupby("Department")["EmployeeCount"].count().idxmin()
a5=a.groupby("Department")["EmployeeCount"].count().min()
print("*Department which have minimum number of employees is :\n",a4,"\n*Number of employees of that Department are :\n",a5)

# Finding Total Number of Employee Attrition
c = a["Attrition"].value_counts()["Yes"]
print("*Total Number of Employee Attrition are : ", c)

# Finding Department Wise Number of Employee Attrition
c1 = a[a["Attrition"] == "Yes"].groupby("Department")["Attrition"].count()
print("*Below are the Departement Wise Number of Employee Attriton : \n", c1)

# Finding Gender Wise Number of Employees
d = a.groupby("Gender")["EmployeeCount"].count()
print("*Below are Gender Wise Number of Employees : \n",d)

# Finding Gender Wise Number of Employee Attrition
d1 = a[a["Attrition"] == "Yes"].groupby("Gender")["Attrition"].count()
print("*Below are the Gender wise Number of Employee Attrition : \n",d1)

# Finding Number of Employees on behalf of Business Travel
e = a.groupby("BusinessTravel")["EmployeeCount"].count()
print("*Below are the Number of Employees based on Business Travel : \n", e)

# Finding Number of Employee Attrition based on Business Travel
e1 = a[a["Attrition"] == "Yes"].groupby("BusinessTravel")["Attrition"].count()
print("*Below are the Number of Employee Attrition based on Business Travel : \n",e1)

# Finding Employee Attrition Rate
f = a["Attrition"].value_counts()["Yes"]
f1 = a["EmployeeCount"].count()
f2 = f/f1*100
print("*Employee Attrition Rate is : ", f2)

# Overtime vs Employee Attrition
g = a[a["Attrition"] == "Yes"].groupby("OverTime")["Attrition"].count()
print("** Overtime Vs Employee Attrition :\n", g)
# Key Outcome is that Overtime is a Factor for Employee Attrition

# Monthly Salary vs Employee Attrition
h = a[a["Attrition"] == "Yes"].groupby(a["MonthlyIncome"] >= 10000)["Attrition"].count()
print("** Number of Employees Attrition with Monthly Income greater than or equal to 10000 represent True and Less than 10000 represent False :\n"
      ,h)
# Key Outcome is that Employees Who have monthly Income less than 10000 are leaving company.

# Total Active Employees
x1 = a["EmployeeCount"].count()
x2 = a["Attrition"].value_counts()["Yes"]
x3 = x1 - x2
print("*Total Active Employees are : ", x3)

# Average Employee Age 
x4 = a["EmployeeCount"].count()
x5 = a["Age"].sum()
x6 = x5/x4
print("*Average Employee Age is : ", x6)

# Distance From Home vs Employee Attrition
k = a[a["Attrition"] == "Yes"].groupby(a["DistanceFromHome"] >= 15)["Attrition"].count()
print("** Number of Employees Attrition with distance from home is equal to or greater than 15kms represent True and less than represent False : \n"
      ,k)
# Key Outcome is that Distance from Home is not a factor for Employee Attrition.
```

```python
# Data Visualization
import matplotlib.pyplot as plt
import pandas as pd
a = pd.read_csv('Employee Attrition.csv')

# Visualizing Department wise Number of Employees
x = a["Department"].value_counts()
x.plot(kind = 'bar',color = "Grey",edgecolor = "Yellow")
plt.title("Department Wise Number of Employees",color = "Purple")
plt.xlabel("Department",color = "Blue")
plt.ylabel("Number of Employees")
plt.show()

# Visualizing Department wise Number of Employees Attrition
y = a[a["Attrition"] == "Yes"].groupby("Department")["Attrition"].count()
y.plot(kind = "line",marker = "*")
plt.title("Department Wise Employee Attrition",color = "Purple")
plt.xlabel("Department",color = "Blue")
plt.ylabel("Number of Employee Attrition")
plt.show()

# Visualizing Gender Wise Number of Employees
d = a.groupby("Gender")["EmployeeCount"].count()
d.plot(kind = "area")
plt.title("Gender Wise Number of Employees",color = "Purple")
plt.ylabel("Number of Employees")
plt.show()

# Visualizing Gender Wise Employee Attrition
d1 = a[a["Attrition"] == "Yes"].groupby("Gender")["Attrition"].count()
d1.plot(kind="pie",legend = "True")
plt.title("Gender Wise Employee Attrition",color = "Purple")
plt.show()

# Viaualizing Number of Employees Based on Business Travel
e = a["BusinessTravel"].value_counts()
e.plot(kind="bar")
plt.title("Number of Employees Based on Business Travel",color="Purple")
plt.xlabel("Business Travel",color = "Grey")
plt.ylabel("Number of Employees")
plt.show()

# Visualizing Employee Attrition Based on Business Travel
e1 = a[a["Attrition"] == "Yes"].groupby("BusinessTravel")["Attrition"].count()
e1.plot(kind="barh",color = "Red")
plt.title("Employee Attrition Based on Business Travel",color = "Purple")
plt.ylabel("Business Travel",color="Grey")
plt.xlabel("Employee Attrition")
plt.show()

```
